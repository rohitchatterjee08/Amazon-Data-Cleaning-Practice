```python
# Check the dataframe columns and their data types using `df.columns` and `df.dtypes`.

print(df.columns)
df.dtypes
```

```python
# Remove unnecessary columns from the data frame.
df.drop(columns= ["img_link", "product_link", 'review_id','user_name','user_id','about_product','product_id'], inplace = True)
```

```python
# Remove the currency symbols and commas from `discounted_price` and `actual_price`, then convert them to numeric values.
# Clean the price columns: remove '₹' and ',' and convert to float.

df["discounted_price"] = df["discounted_price"].str.replace("₹", "").str.replace(",", "").astype(float)

df["actual_price"] = df["actual_price"].str.replace("₹", "").str.replace(",", "").astype(float)
```

```python
df = df[df["rating"] != "|"]
```
